```python
# https://docs.google.com/spreadsheets/d/19a_zZzf1h7Li_QfTB3TV49jjbTSSL8xdABgc9meiRjY/edit?usp=sharing

import pandas as pd
import requests
import time

endpoint_planilha = "https://docs.google.com/spreadsheets/d/19a_zZzf1h7Li_QfTB3TV49jjbTSSL8xdABgc9meiRjY/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print('----- Dados Coletados -----')
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

```python
# iterar cada resposta do formulario
for indice, linha in dados_planilha.iterrows():
  nome_local = linha['Nome do Local']
  lat = linha['Latitude']
  lon = linha['Longitude']

  parametros = {
      'latitude':lat,
      'longitude':lon,
      'current':'temperature_2m'
  }
  print(f'-----{nome_local}-----')
  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)
  print('\n\n')
```
